#**Graph 1: Monthly Sales and Profit Trend**

```python

df["Month"] = df["Date"].dt.to_period("M").astype(str)


monthly_data = df.groupby("Month")[["Sales", "Profit"]].sum()


plt.figure(figsize=(10, 5))

plt.plot(monthly_data.index,
         monthly_data["Sales"],
         marker="o",
         label="Sales",linestyle=":")

plt.plot(monthly_data.index,
         monthly_data["Profit"],
         marker="o",
         label="Profit",
         linestyle=":")

plt.title("Monthly Sales and Profit Trend",fontweight='bold')
plt.xlabel("Month",fontweight='bold')
plt.ylabel("Amount",fontweight='bold')
plt.xticks(rotation=0)
plt.style.use("ggplot")
plt.grid(True)
plt.legend()

plt.show()
```

#**Graph 2: Sales by Category**

```python
# Calculate total sales for each category
category_sales = df.groupby("Category")["Sales"].sum()

# Create bar chart
plt.figure(figsize=(8, 5))

plt.bar(category_sales.index, category_sales.values,color="seagreen")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

#**Graph 3: Sales by Region**

```python
# Calculate total sales for each region
region_sales = df.groupby("Region")["Sales"].sum()

# Create bar chart
plt.figure(figsize=(8, 5))

plt.bar(region_sales.index, region_sales.values,color="seagreen")

plt.title("Sales by Region")
plt.xlabel("Region",fontweight="bold")
plt.ylabel("Total Sales",fontweight="bold")

plt.show()
```

#**Graph 4: Sales by Product**

```python
# Calculate total sales for each product
product_sales = df.groupby("Product")["Sales"].sum()

# Sort values from highest to lowest
product_sales = product_sales.sort_values(ascending=False)

# Create horizontal bar chart
plt.figure(figsize=(10, 6))

plt.barh(product_sales.index, product_sales.values,color="seagreen")

# Put highest product at the top
plt.gca().invert_yaxis()

plt.title("Sales by Product")
plt.xlabel("Total Sales")
plt.ylabel("Product")

plt.show()
```

```python


# Load dataset
df = pd.read_csv("/sales_data.csv")

# Convert Date to datetime
df["Date"] = pd.to_datetime(df["Date"])

# Create Month column
df["Month"] = df["Date"].dt.to_period("M").astype(str)

# Monthly Sales and Profit
monthly_data = df.groupby("Month")[["Sales", "Profit"]].sum()

# Sales by Category
category_sales = df.groupby("Category")["Sales"].sum()

# Sales by Region
region_sales = df.groupby("Region")["Sales"].sum()

# Sales by Product
product_sales = df.groupby("Product")["Sales"].sum()
product_sales = product_sales.sort_values(ascending=False)


# ==========================================
# Create ONE figure with 4 graphs
# ==========================================

plt.figure(figsize=(16, 10))


# -------------------------------
# 1. Monthly Sales and Profit
# -------------------------------
plt.subplot(2, 2, 1)

plt.plot(monthly_data.index,
         monthly_data["Sales"],
         marker="o",
         label="Sales",linestyle=":")

plt.plot(monthly_data.index,
         monthly_data["Profit"],
         marker="o",
         label="Profit",linestyle=":")

plt.title("Monthly Sales and Profit Trend")
plt.xlabel("Month",fontweight="bold")
plt.ylabel("Amount",fontweight="bold")
plt.xticks(rotation=90)
plt.legend()
plt.grid(True)


# -------------------------------
# 2. Sales by Category
# -------------------------------
plt.subplot(2, 2, 2)

plt.bar(category_sales.index,
         category_sales.values,color="seagreen")

plt.title("Sales by Category")
plt.xlabel("Category",fontweight="bold")
plt.ylabel("Total Sales",fontweight="bold")
plt.grid(False)


# -------------------------------
# 3. Sales by Region
# -------------------------------
plt.subplot(2, 2, 3)

plt.bar(region_sales.index,
         region_sales.values,color="yellow")

plt.title("Sales by Region")
plt.xlabel("Region",fontweight="bold")
plt.ylabel("Total Sales",fontweight="bold")
plt.grid(False)


# -------------------------------
# 4. Sales by Product
# -------------------------------
plt.subplot(2, 2, 4)

plt.barh(product_sales.index,
         product_sales.values,color="brown")

plt.gca().invert_yaxis()

plt.title("Sales by Product")
plt.xlabel("Total Sales",fontweight="bold")
plt.ylabel("Product",fontweight="bold")


# Adjust spacing
plt.tight_layout()
plt.style.use("ggplot")
plt.grid(True)

# Display all 4 graphs together
plt.show()
```

#**Sales Data-analysis Dashboard**

```python

# Create a 2 × 2 subplot
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# -------------------------------------------------
# 1. PIE CHART - Sales Distribution by Category
# -------------------------------------------------

category_sales = df.groupby("Category")["Sales"].sum()

colors = [
    "#FF6B6B",
    "#4ECDC4",
    "#FFD166",
    "#6C5CE7",
    "#00B894"
]

explode = [0.06] * len(category_sales)

axes[0, 0].pie(
    category_sales,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    colors=colors[:len(category_sales)],
    shadow=True,
    pctdistance=0.50,
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1.5
    },
    textprops={
        "fontsize": 10,
        "fontweight": "bold"
    }
)

axes[0, 0].set_title(
    "Sales Distribution by Category",
    fontsize=15,
    fontweight="bold"
)

# -------------------------------------------------
# 2. HISTOGRAM - Sales Distribution
# -------------------------------------------------

axes[0, 1].hist(
    df["Sales"],
    bins=10,
    edgecolor="black",
    linewidth=1,
    color="seagreen",
    alpha=0.75
)

axes[0, 1].set_title(
    "Sales Distribution – 120 Transactions",
    fontsize=15,
    fontweight="bold"
)

axes[0, 1].set_xlabel(
    "Sales Amount (₹)",
    fontsize=11,
    fontweight="bold"
)

axes[0, 1].set_ylabel(
    "Frequency",
    fontsize=11,
    fontweight="bold"
)

axes[0, 1].grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

# -------------------------------------------------
# 3. HISTOGRAM - Unit Price Distribution
# -------------------------------------------------

axes[1, 0].hist(
    df["Unit_Price"],
    bins=10,
    edgecolor="black",
    linewidth=1,
    color="teal",
    alpha=0.75
)

axes[1, 0].set_title(
    "Distribution of Unit Price",
    fontsize=15,
    fontweight="bold"
)

axes[1, 0].set_xlabel(
    "Unit Price (₹)",
    fontsize=11,
    fontweight="bold"
)

axes[1, 0].set_ylabel(
    "Frequency",
    fontsize=11,
    fontweight="bold"
)

axes[1, 0].grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

# -------------------------------------------------
# 4. SCATTER PLOT - Sales vs Profit
# -------------------------------------------------

scatter = axes[1, 1].scatter(
    df["Sales"],
    df["Profit"],
    s=df["Quantity"] * 8,
    alpha=0.6,
    c=df["Quantity"],
    cmap="viridis",
    edgecolors="black",
    linewidths=0.6
)

axes[1, 1].set_title(
    "Sales vs Profit",
    fontsize=15,
    fontweight="bold"
)

axes[1, 1].set_xlabel(
    "Sales (₹)",
    fontsize=11,
    fontweight="bold"
)

axes[1, 1].set_ylabel(
    "Profit (₹)",
    fontsize=11,
    fontweight="bold"
)

axes[1, 1].grid(
    True,
    linestyle="--",
    alpha=0.3
)

# Add colorbar for Quantity
cbar = fig.colorbar(
    scatter,
    ax=axes[1, 1]
)

cbar.set_label(
    "Quantity",
    fontsize=10,
    fontweight="bold"
)

# -------------------------------------------------
# Overall title
# -------------------------------------------------

fig.suptitle(
    "Sales Data Analysis Dashboard",
    fontsize=22,
    fontweight="bold",
    y=0.98
)

# Adjust spacing
plt.tight_layout(rect=[0, 0, 1, 0.95])

# Display
plt.show()
```
